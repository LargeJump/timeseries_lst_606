We can test to see if the area of statistically significant hotspots has changed  accross each of the epochs using hypothesis testing. Each count of a value represents a pixel with an area of $900m^2$. The hotspots are considered to be significant to 99% or p = 0.01. This data falls within the $Gi Bin = 3$. 

We test the hypothesis in the following cells. 

 ($H_0: \text{Area}_{1996} = \text{Area}_{2006} = \text{Area}_{2017}$)

```python
import numpy as np
from scipy import stats

# Assume `area_matrix` is a 2D NumPy array of shape (n_spatial_units, 3)
# Columns: [Area_1996, Area_2006, Area_2017]
# Example structure:
# area_matrix = active_blocks[["HS_lst_96", "HS_lst_06", "HS_lst_17"]].to_numpy()

stat, p_val = stats.friedmanchisquare(
    [53067],  # 1996
    [53071],  # 2006
    [51360]   # 2017
)

print(f"Friedman Test (H0: A = B = C): chi2 = {stat:.4f}, p = {p_val:.4e}")

if p_val < 0.05:
    print("Result: Reject H0 — hotspot area differs significantly across epochs.")
else:
    print("Result: Fail to reject H0 — no statistically significant change.")
```
